```sql
SELECT
    COUNT(*) AS total_rows,
    SUM(CASE
        WHEN Quantity IS NULL OR TRIM(Quantity) = '' THEN 1
        ELSE 0
    END) AS blank_rows,
    SUM(CASE
        WHEN Quantity IS NOT NULL
         AND TRIM(Quantity) <> ''
         AND TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) IS NULL THEN 1
        ELSE 0
    END) AS unparseable_nonblank_rows,
    SUM(CASE
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) IS NOT NULL
         AND TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4))
             <> FLOOR(TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4))) THEN 1
        ELSE 0
    END) AS fractional_rows,
    SUM(CASE WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) > 0 THEN 1 ELSE 0 END) AS positive_count,
    SUM(CASE WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) < 0 THEN 1 ELSE 0 END) AS negative_count,
    SUM(CASE WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) = 0 THEN 1 ELSE 0 END) AS zero_count,
    MIN(TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4))) AS min_quantity,
    MAX(TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4))) AS max_quantity
FROM workspace.retail_project.bronze_online_retail_raw;
```

```sql
SELECT
    COUNT(*) AS total_rows,
    SUM(CASE
        WHEN UnitPrice IS NULL OR TRIM(UnitPrice) = '' THEN 1
        ELSE 0
    END) AS blank_rows,
    SUM(CASE
        WHEN UnitPrice IS NOT NULL
         AND TRIM(UnitPrice) <> ''
         AND TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) IS NULL THEN 1
        ELSE 0
    END) AS unparseable_nonblank_rows,
    SUM(CASE
        WHEN TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) IS NOT NULL
         AND TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4))
             <> FLOOR(TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4))) THEN 1
        ELSE 0
    END) AS fractional_rows,
    SUM(CASE WHEN TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) > 0 THEN 1 ELSE 0 END) AS positive_count,
    SUM(CASE WHEN TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) < 0 THEN 1 ELSE 0 END) AS negative_count,
    SUM(CASE WHEN TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) = 0 THEN 1 ELSE 0 END) AS zero_count,
    MIN(TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4))) AS min_unit_price,
    MAX(TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4))) AS max_unit_price
FROM workspace.retail_project.bronze_online_retail_raw
```

```sql
SELECT 
    CASE 
        WHEN InvoiceNo IS NOT NULL 
        AND UPPER(TRIM(InvoiceNo)) LIKE 'C%'
        THEN 'C-prefixed'
        ELSE 'not C-prefixed'
    END AS invoice_class,
    CASE
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) > 0 THEN 'POSITIVE'
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) < 0 THEN 'NEGATIVE'
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) = 0 THEN 'ZERO'
        ELSE 'unparsed/blank'
    END AS quantity_class,
    COUNT(*) AS line_count
FROM workspace.retail_project.bronze_online_retail_raw
GROUP BY CASE 
        WHEN InvoiceNo IS NOT NULL 
        AND UPPER(TRIM(InvoiceNo)) LIKE 'C%'
        THEN 'C-prefixed'
        ELSE 'not C-prefixed'
    END ,
    CASE
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) > 0 THEN 'POSITIVE'
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) < 0 THEN 'NEGATIVE'
        WHEN TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) = 0 THEN 'ZERO'
        ELSE 'unparsed/blank'
    END
ORDER BY invoice_class, quantity_class
```

```sql
WITH target AS(
    SELECT TRIM(StockCode) AS StockCode
    FROM workspace.retail_project.bronze_online_retail_raw
    WHERE (InvoiceNo IS NULL OR UPPER(TRIM(InvoiceNo)) NOT LIKE 'C%')
    AND TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) < 0
    AND TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) = 0
    AND Description IS NULL
    AND CustomerID IS NULL
),
described AS(
    SELECT TRIM(StockCode) AS StockCode,
    COUNT(DISTINCT TRIM(Description)) AS description_variants,
    MIN(TRIM(Description)) AS example_description
    FROM workspace.retail_project.bronze_online_retail_raw
    WHERE Description IS NOT NULL AND TRIM(Description) <> ''
    GROUP BY TRIM(StockCode)
)
SELECT
    COUNT(*) AS target_lines,
    COUNT(DISTINCT TRIM(t.StockCode)) AS distinct_target_stocks,
    SUM(CASE WHEN d.StockCode IS NOT NULL THEN 1 ELSE 0 END)
        AS lines_with_description_elsewhere,
    COUNT(DISTINCT CASE WHEN d.StockCode IS NOT NULL THEN t.StockCode END)
        AS codes_with_description_elsewhere,
    COUNT(DISTINCT CASE WHEN d.StockCode IS NULL THEN t.StockCode END)
        AS codes_without_any_description

    FROM target t
    LEFT JOIN described d
    ON t.StockCode = d.StockCode;
```

```sql
WITH target AS(
    SELECT TRIM(StockCode) AS StockCode
    FROM workspace.retail_project.bronze_online_retail_raw
    WHERE (InvoiceNo IS NULL OR UPPER(TRIM(InvoiceNo)) NOT LIKE 'C%')
    AND TRY_CAST(TRIM(Quantity) AS DECIMAL(18, 4)) < 0
    AND TRY_CAST(TRIM(UnitPrice) AS DECIMAL(18, 4)) = 0
    AND Description IS NULL
    AND CustomerID IS NULL
),
target_codes AS(
    SELECT DISTINCT StockCode
    FROM target
),
variants AS(
    SELECT TRIM(StockCode) AS StockCode,
    COUNT(DISTINCT TRIM(Description)) AS description_variants
    FROM workspace.retail_project.bronze_online_retail_raw
    WHERE Description IS NOT NULL AND TRIM(Description) <> ''
    GROUP BY TRIM(StockCode)
)
SELECT
    COUNT(*) AS distinct_target_codes,
    SUM(CASE WHEN v.description_variants = 1 THEN 1 ELSE 0 END) AS codes_with_one_description,
SUM(CASE WHEN v.description_variants > 1 THEN 1 ELSE 0 END) AS codes_with_multiple_descriptions,
SUM(CASE WHEN v.StockCode IS NULL THEN 1 ELSE 0 END) AS codes_without_description
FROM target_codes t
LEFT JOIN variants v
ON t.StockCode = v.StockCode;
```
